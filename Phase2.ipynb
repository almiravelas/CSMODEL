### **Survey on Overseas Filipinos 2015**
Last October of 2015, the Philippine Statistics Authority (PSA) created a dataset on a Survey on Overseas Filipinos (SOF). The survey was designed to gather data on the amount of overseas Filipino workers (OFW) along with their socio economic characteristics and other information related to working abroad from April to September. The remittances of the OFWs were also taken into account for the specified period. 


### **Collection Process**
**Sampling Procedure** <br>
**The SOF used the sampling design of the 2003 Master Sample for Household Surveys.*
1. `Domain` - **each of the country's 17 administrative regions** were considered as a domain for this survey
2. `Sampling Frame` - the primary sampling units (PSU) were defined as a **barangay or a combination of barangays with at least 500 households**
3. `Sample Size` - **2,835 PSUs** (330 certainty PSUs, 2,505 non-certainty PSUs)
4. `Stratification` - the **17 regions** of the country were considered as the primary strata
5. `Sample Selection` - **divided into 4 sub-samples** with probability proportional to some estimated measure of size of the total number of households from the 2000 CPH

**Data Collection**
- `Dates of Collection` - 2015-10-08 (October 8, 2015) - 2015-10-30  (October 8, 2015)
- `Mode of Data Collection` - Face-to-face (F2F)
- `Data Collection Supervision` - **supervisors** are required to have an itinerary of travel for efficiency and security, a copy of the said itinerary is also available in the field office in case of emergencies. Supervisors are also the ones who are tasked with taking action in any problems that may arise hence regional directors and provincial statistical officers are encouraged to visit them to rate their progress.
- `Questionnaires` - **SOF Form 2** is the household (HH) control form and is the basis to determine households with members that is overseas. HHs with overseas members in the specified time period were given the SOF Form 1. The other questionnaire, SOF Form 1, is designed to gather data on the number of overseas Filipinos, their socio-economic characteristics, and their remittances.
- `Data Collectors` - Philippine Statistics Authority (PSA)
  


### **Data Collection Implications**
The data collection implies that the resulting conclusions and insights will be related to `socio-economic characteristics, the migration patterns of OFWs, and the OFWs' remittance behaviors`.<br>
There are some variables that are directly defines OFWs' socio-economic characteristics like RQ8_HGRADE (Highest Grade Completed) while some like RQ19_REASON (Reason for returning in the country) may suggest or prove their current standing.<br>
As for the migration patterns, variables starting with RQ10 to RQ20 are related to going abroad and their length of stay overseas. It also includes the reasons for leaving or returning to the country, the expected and actual dates of returning home, the kind of work abroad, and how long they worked for overseas.<br>
Lastly, for the remittance behaviors, variables starting with RQ21 to RQ24 are all remittance-related with 2 variables stating the amount received, the mode of remittance, and the amounts of remittance spent on certain categories. 


### **Dataset Structure**
The survey on overseas Filipinos' data structure contains the overseas Filipino (OF) identifier and his or her data related to the survey. There are exactly `5440 observations` in the dataset each representing `1 OF which is placed per row` of the data frame. Each row contains their personal information, information about leaving or staying abroad, and data on their remittances. On the other hand, `*each column represents the values of the OFs per variable`. This particular dataset contains `48 variables`, 2 of which are used as identifiers, namely RREG (Region) and HHNUM (Unique HH Serial Number).

# *1.1 Preprocessing*

Before running the K-means clustering algorithm, the dataset still needs a few additional adjustments. This includes setting some values back to integers so that they are compatible with the algorithm. Among the variables encoded back to integers are `RQ8_HGRADE`, `RQ3_SEX`, and `RQ7_MSTAT`.

In remapping the highest grade variable `RQ8_HGRADE`, certain integer values were set to nudge the clustering algorithm to consider wider gaps between some categories. For example, it would consider a high school undergraduate who dropped out to be closer to an elementary graduate compared to a high school graduate. In addition, Tech/Voc and College were merged in this encoding and were given the same values: 30 and 45.

```python
# generate new partial frame for sklearn's k-means
k_frame_full = pd.DataFrame(data).dropna(subset=["RQ22_CASHAMT"])
k_frame_full.loc[k_frame_full["RQ13_STAY"] == np.inf, "RQ13_STAY"] = 500 # set to arbitrarily high value

# map this to an integer format
# undergraduate > graduate = +10
# graduate > next level = +5
# techvoc/college > graduation = +15
k_frame_full["HGRADE_INT"] = k_frame_full["RQ8_HGRADE"].map({
    "Elementary undergraduate": "0",
    "Elementary graduate": "10",
    "High school undergraduate": "15",
    "High school graduate": "25",
    "Technical-vocational undergraduate": "30",
    "Technical-vocational graduate": "45",
    "College undergraduate": "30",
    "College graduate": "45",
    None: "0",
}).astype("string").fillna("0").astype("int64")
```

Mapping sex, `RQ3_SEX`, to an integer is considerably simpler. `Male` was set to the value `-5` and `Female` was set to the value `5`, thus providing a difference of 10 units for people with different genders. Mapping a categorical value here is easy as the column only contains two unique values and thus can be mapped equidistantly under one axis; if an `Agender` or an otherwise nonbinary value existed in this column, it may require a One Hot Encoding approach to be fully mapped here.

```python
# map sex to integers
k_frame_full["SEX_INT"] = k_frame_full["RQ3_SEX"].map({
    "Male": -5,
    "Female": 5,
    None: 0,
}).astype("string").fillna("0").astype("int64")
```

Mapping marriage status, `RQ7_MSTAT`, would fall into the same problem as above as this column has five potential values (excluding `Unknown`). However, we decided to make this column simpler instead by considering "has this person married?" instead as marriage is usually linked to the presence of children or other dependents in the person's life.

```python
# map marital status to integers
# set to a boolean question "Has [x] married?" regardless of current status of marriage
k_frame_full["MARRIED_INT"] = k_frame_full["RQ7_MSTAT"].map({
    "Single": "0",
    "Married": "10",
    "Widowed": "10",
    "Separated": "10",
    "Common-law": "0",
    "Unknown": 0,
    None: "0",
}).astype("string").astype("int64")
```

# *1.2 Applying Modelling Technique*

With specialized data cleaning done once again, we can now begin the clustering process. This block retrieves the columns that clustering would consider and then runs the clustering algorithm.

```python
# get columns
k_frame = pd.DataFrame(k_frame_full, columns = ["RQ4_AGE", "RQ16_NOMONTH", "SEX_INT", "HGRADE_INT", "MARRIED_INT"]).values
```

Silhouette Score: This measures how well-defined clusters are. A high score means that data points are closely related to others in their cluster and very different from points in other clusters.

Inertia: This measures how tightly packed data points are within each cluster. A low inertia means that points are close to their cluster's center.

By looking at both scores, we can find the best number of clusters. We want clusters that are both well-separated (high silhouette score) and tightly packed (low inertia)

```python
# Calculate silhouette scores for different k values
scores = []
inertia = []
k_values = range(2, 30)

for k in k_values:
    km_test = KMeans(n_clusters=k, random_state=122).fit(k_frame)
    scores.append(silhouette_score(k_frame, km_test.labels_))
    inertia.append(km_test.inertia_)

# Plot silhouette scores
plt.figure(figsize=(10, 5))
plt.plot(k_values, scores, marker='o')
plt.title("Silhouette Score vs. Number of Clusters (k)")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.show()
```

In this graph, the silhouette score starts high for fewer clusters and decreases as the number of clusters increases.
The silhouette score rapidly decreases after 3 clusters, suggesting that clustering beyond this may lead to overlapping or poorly separated clusters.

```python

# Plot inertia
plt.figure(figsize=(10, 5))
plt.plot(k_values, inertia, marker='o')
plt.title("Inertia vs. Number of Clusters (k)")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.show()
```

As the number of clusters increases, inertia decreases significantly, but the rate of decrease slows after around 3–5 clusters. This indicates diminishing returns in adding more clusters.

```python
# run kmeans !
n_clusters = 4
random_state = 122
km = KMeans(n_clusters = n_clusters, random_state = random_state)
km.fit(k_frame)
```

# Statistical Inferencing

**Null Hypothesis (Ho)**: OFWs of similar demographics and backgrounds do not remit similar amounts of money.

**Alternative Hypothesis (Ha)**: OFWs of similar demographics and backgrounds remit similar amounts of money back to their families.

Data preprocessing, preparing the data for T-testing.

We will use `Log_CashRemittances`, a log-transformed version of `RQ22_CASHAMT`, as the main column for this test.

```python
from scipy.stats import f_oneway

# Group data by clusters
cluster_data = [k_frame_full[km.labels_ == i]["RQ22_CASHAMT"] for i in range(n_clusters)]

# Filter out empty groups (if any)
cluster_data = [group.dropna() for group in cluster_data if len(group) > 0]
```

We use visualization diagrams like histograms to validate if the ``RQ22_CASHAMT`` are normally distributed within each cluster.

If we analyze the different diagrams and see that the histogram is bell-shaped and symmetrical, then the data within the cluster is approximately normally distributed. If it leans to the left (negative skew) or right (positive skew), the data is not symmetric. Whilst, multiple peaks suggests that the cluster may contain subgroups with different characteristics.

```python
for i in range(n_clusters):
    cluster_data_temp = k_frame_full[km.labels_ == i]["Log_CashRemittances"].dropna()
    if len(cluster_data_temp) > 0:
        plt.figure(figsize=(8, 4))
        sns.histplot(cluster_data_temp, kde=True, bins=20)
        plt.title(f"Cluster {i} - Histogram and KDE")
        plt.xlabel("Amount remitted to family (log-transformed)")
        plt.ylabel("Frequency")
        plt.show()
```

For example, Cluster 0:
- If the histogram shows a single, symmetric peak, the cluster members tend to have consistent normalized remittance values.
- If the KDE (Kernel Density Estimation) curve has two peaks, there might be subgroups with differing behaviors within this cluster.
- If the histogram is heavily skewed, it suggests most members of this cluster either remit low or high amounts relative to others.

We can now run the T-test. This one-tailed T-test compares a series of all standard errors from each cluster and determines whether this series is significantly different from the population's standard deviation.

```python
agg_per_cluster["std"]
res = stats.ttest_1samp(agg_per_cluster["std"], k_frame_full["RQ22_CASHAMT"].std(), alternative="less")

print(f"F-Statistic: {res.statistic:.4f}")
print(f"P-Value: {res.pvalue:.4f}")
if res.pvalue < 0.05:
    print("Reject the null hypothesis, clusters of similar demographics remit similar amounts of money.")
else:
    print("Fail to reject the null hypothesis, clusters of similar demographics do not necessarily remit similar amounts of money.")
```

After testing, the f-score is **-2.0431**. Moreover, the resulting p-val is **0.0251**.

Since these p-values are less than the significance level of 0.05, **we reject the null hypothesis**.

Therefore, we conclude that **clusters of similar demographics remit similar amounts of money**.
